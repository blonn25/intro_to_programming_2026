## 1.1 What is an Array?

Numpy arrays are more efficient than Python lists for numerical operations. Let's create a simple 1D array and check its basic attributes.

````python
import numpy as np

mylist = [2, 5, 3, 9, 5, 2]
myarray = np.array(mylist)

print(type(mylist))
print(mylist)
print()
print(type(myarray))  # <class 'numpy.ndarray'>
print(myarray.dtype)  # int64
print(myarray.shape)  # (6,)
print(myarray)
````

````python
two_dimensional_list = [
    [1, 2, 3],
    [4, 5, 6]
]
print(two_dimensional_list)

two_dimensional_array = np.array(two_dimensional_list)
print(two_dimensional_array.shape)
print(two_dimensional_array)
````

### Bonus - How to edit an array

````python
# edit values in an arrray
demo_array = np.array(
    [[1, 2, 3],
     [4, 5, 6]]
)
print(demo_array)

demo_array[1, 1] = 30
print(demo_array)
````

## 1.2 Creating Arrays

Numpy has several built-in ways to generate commonly used arrays.

````python
zeros_array = np.zeros((2, 3))  # np.zeros(other_array.shape)
ones_array = np.ones((2, 3))
range_array = np.arange(0, 10, 2)
linspace_array = np.linspace(0, 1, 5)

print(zeros_array)
print(ones_array)
print(range_array)
print(linspace_array)
````

## 2.1 Element-wise Operations

Math operations are automatically applied to every element in the array.

````python
print(myarray + 3)
print(myarray * 2)
````

## 2.2 Mathematical Functions

Numpy has built-in functions for common math operations.

````python
print(np.sin(myarray))
print(np.exp(myarray))
print(np.log10(myarray))
````

## 2.3 Logical Operations

Comparisons on an array return a boolean array, element by element.

````python
print(myarray)
print(myarray > 3)
````

## 2.4 Array Aggregation

Functions like `sum`, `mean`, and `std` can be applied across the whole array or along a specific `axis`.

````python
nd_array = np.random.normal(10, 2, (3, 4))
print(nd_array)
print(nd_array.sum())          # Total sum
print(nd_array.mean())         # Total Mean
print(nd_array.mean(axis=0))   # Mean along the first axis (rows)
print(nd_array.mean(axis=1))
````

## 3.1 Line Plot

Let's use numpy to generate data and matplotlib to plot it: a sine wave and a noisy version of the same wave.

````python
import matplotlib.pyplot as plt

time = np.arange(0, 10, 0.1)
sine_wave = np.sin(time)
noisy_wave = sine_wave + np.random.normal(0, 0.2, sine_wave.shape)

plt.plot(time, sine_wave)
plt.plot(time, noisy_wave, 'o-')
plt.xlabel('Time')
plt.ylabel('Amplitude')
plt.title('Sine Wave with Noise')
plt.show()
````

## 3.2 Histogram Plot

A histogram shows the distribution of values in an array.

````python
plt.hist(noisy_wave, bins=10)
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Histogram of Noisy Sine Wave')
plt.show()
````

## 3.3 Image Plot (2D Array)

`imshow()` visualizes a 2D array as an image, mapping values to colors.

````python
random_array = np.random.rand(5, 5)
print(random_array)

plt.imshow(random_array, cmap='viridis')
plt.colorbar()
plt.title('Random 2D Array')
plt.show()
````
