## Step 1: DNA → RNA (transcription)

Transcription replaces every `T` with `U`. This is a single string method call — try it on a small example below.

```python
example_dna = "ATGAGC"
example_rna = example_dna.replace('T', 'U')
print(example_rna)
```

## Step 2: Splitting RNA into codons

Translation reads RNA three letters (one codon) at a time. Below is a *generic* demonstration of walking through a string in chunks of 3 — notice this uses the same slicing (`[start:stop]`) and `while`-loop patterns from Day 3.

```python
example_string = "ABCDEFGHI"  # stand-in for an RNA sequence, 3 chars = one "codon"

start_position = 0
while start_position < len(example_string):
    chunk = example_string[start_position:start_position + 3]
    print(chunk)
    start_position += 3
```

Try applying that same chunking pattern to `example_rna` from Step 1, and look each resulting codon up in `genetic_code`.

```python
start_position = 0
while start_position < len(example_rna):
    codon = example_rna[start_position:start_position + 3]
    print(codon, "->", genetic_code[codon])
    start_position += 3
```

## Step 3: Reading a sequence from a FASTA file

Recall from Day 3 how to open and read a file. A FASTA file's first line is a `>`-prefixed header, and the sequence follows on the next line(s).

```python
with open('mystery_protein.fa', 'w') as f:
    f.write(">mystery_protein_seq\n")
    f.write("ATGGACAAGGACGAGGCGTGGAAGTGCGTGGAGCAGCTGAGG\n")
```

```python
with open('mystery_protein.fa', 'r') as f:
    lines = f.readlines()

header = lines[0]
sequence = lines[1].strip()
print("Header: ", header.strip())
print("Sequence:", sequence)
```
