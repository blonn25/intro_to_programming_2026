## 3.1 Introduction to Lists

Lists are mutable sequences that can hold various data types.

````python
cats = ['Tom', 'Snappy', 'Kitty', 'Jessie', 'Chester']
cats
````

You can add, remove, or change items in a list:

````python
cats.append('Catherine')  # Adds a new cat to the list
print(cats)
````

````python
del cats[1]  # Removes 'Snappy' from the list
print(cats)
````

````python
cats[0] = 'Smores'  # Changes the first name 'Tom' to 'Smores'
print(cats)
````

````python
cats.remove('Jessie')  # Removes 'Jessie' from the list
print(cats)
````

## 3.2 For Loops

The `for` loop iterates over a sequence (like a list) and performs an action for each item.

````python
newList = [45, 'eat me', 90210, "The day has come", -67]

for value in newList:
    #value = 'example'
    print(value)

print(newList)
````

## 4.1 While Loops

The `while` loop runs as long as a condition is true.

````python
a = 0
while a < 10:
    print(a)
    a += 1
````

## 4.2 Conditional Statements

Conditional statements allow code to run only if certain conditions are met.

````python
y = 1
if y == 1:
    print("y equals 1")
````

## 4.3 Boolean Expressions

Boolean expressions evaluate to `True` or `False`:

| Expression | Function                 |
| ---------- | ------------------------ |
| `<`        | Less than                |
| `<=`       | Less than or equal to    |
| `>`        | Greater than             |
| `>=`       | Greater than or equal to |
| `!=`       | Not equal to             |
| `==`       | Equal to                 |

````python
print(3 < 5)
print(3 >= 5)
print(3 != 5)
````

## 4.4 More Complex Conditions

Use `elif` and `else` to handle different conditions.

````python
z = 4
if z > 70:
    print("Something is very wrong")
elif z < 7:
    print("This is normal")
````

````python
x = 5
if type(x) != int:
    print("x is not a number")
elif x > 5:
    print("x is greater than 5")
elif x < 5:
    print("x is less than 5")
else:
    print("x is equal to 5")
````

## 4.5 Indentation in Python

Indentation defines code blocks. Try changing the value of `a` below and re-running to see how the logic branches.

````python
a = 10
while a > 0:
    print(a)
    
    if a > 5:
        print("Big number!")
    elif a % 2 != 0:
        print("This is an odd number")
    else:
        print("Small and even number")

    a -= 1
````

## 5.1 Loading a Text File

We first need something to read — this cell writes a small file into the current folder so the read examples below have something to work with.

````python
with open('example.txt', 'w') as file:
    file.write("Hello, world!")
````

````python
file = open('example.txt')
content_of_file = file.read()
print(content_of_file)
file.close()
````

````python
with open('example.txt', 'r') as file:
    content_of_file = file.read()
    print(content_of_file)
````

The more accepted way to do this is with a **context manager** (`with`), which automatically closes the file for you:

````python
with open('example.txt', 'r') as file:
    content = file.read()
    print(content)
````

## 5.2 Saving a Text File

Write mode (`'w'`) creates the file if it doesn't exist, and overwrites it if it does.

````python
with open('example.txt', 'w') as file:        
    file.write("Hello, world!")
````

## 5.3 File Modes

| Mode | Description                                                                   |
| ---- | ----------------------------------------------------------------------------- |
| 'r'  | Read mode. Opens a file for reading (default).                                |
| 'w'  | Write mode. Opens a file for writing (overwrites existing content).           |
| 'a'  | Append mode. Opens a file for writing (appends to existing content).          |
| 'w+' | Read and write mode. Opens a file for both reading and writing.               |
| 'b'  | Binary mode. Used with other modes (e.g., 'rb', 'wb') to handle binary files. |
